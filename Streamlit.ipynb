```python
import streamlit as st
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import json
import os

st.set_page_config(layout="wide")
st.title("📊 Loan Approval - Random Forest Insights Dashboard")

# Optional: Automatically load from Google Drive if using in Colab
DEFAULT_PATH = "/content/drive/MyDrive/Colab Notebooks/output.txt"

st.sidebar.header("📂 Data Source")
use_default = st.sidebar.checkbox("Use Colab output.txt from Google Drive", value=True)

if use_default and os.path.exists(DEFAULT_PATH):
    with open(DEFAULT_PATH, "r") as f:
        lines = f.readlines()
else:
    avg_file = st.sidebar.file_uploader("Upload avg_age.py Output", type=["txt", "json"])
    lines = avg_file.readlines() if avg_file else None

if lines:
    try:
        # Try reading JSON lines (typical MRJob output)
        results = [json.loads(line) for line in lines if line.strip()]
        df = pd.DataFrame(results, columns=["approved", "avg_age"])
    except:
        # Fallback: try reading plain text CSV-style
        from io import StringIO
        text_data = StringIO("".join([line.decode("utf-8") if isinstance(line, bytes) else line for line in lines]))
        df = pd.read_csv(text_data, names=["approved", "avg_age"])

    st.subheader("📈 Average Age by Loan Approval Status")
    fig, ax = plt.subplots()
    sns.barplot(x="approved", y="avg_age", data=df, palette="viridis", ax=ax)
    ax.set_xlabel("Approved")
    ax.set_ylabel("Average Age")
    st.pyplot(fig)
else:
    st.info("📄 Please upload a file or enable the default path checkbox to load avg_age.py output.")

st.markdown("---")
st.caption("💡 This dashboard reads output from MRJob's avg_age.py script (approved vs average age). It supports both manual file upload and auto-load from Colab Google Drive path.")
```
