# ***MODEL*** **BUILDING**

```python
df.show
```

```python
df.show()
```

```python
from pyspark.ml.feature import VectorAssembler
```

```python
Assembler =VectorAssembler(inputCols=["Pregnancies","Glucose","BloodPressure","SkinThickness","Insulin","BMI","DiabetesPedigreeFunction","Age"],outputCol="features")
```

```python
Assembler
```

```python
df = Assembler.transform(df)
```

```python
df.show(truncate=False)
```

```python
model = df.select("features","Outcome")
```

```python
model.show(truncate = False)
```

```python
train_df, test_df = model.randomSplit([0.8,0.2], seed = 100)
```

```python
train_df.count()
```

```python
test_df.count()
```

LOGISTIC REGRESSION


```python
from pyspark.ml.classification import LogisticRegression
```

```python
lr_model = LogisticRegression(featuresCol="features", labelCol="Outcome")
model_lr = lr_model.fit(train_df)
lr_train_prediction = model_lr.transform(train_df)
lr_test_prediction = model_lr.transform(test_df)
```

```python
lr_train_prediction.show()
```

Decision Tree


```python
from pyspark.ml.classification import DecisionTreeClassifier
```

```python
DT = DecisionTreeClassifier(featuresCol="features", labelCol="Outcome")
model_DT = DT.fit(train_df)
DT_train_prediction = model_DT.transform(train_df)
DT_test_prediction = model_DT.transform(test_df)
```

Random Forests


```python
from pyspark.ml.classification import RandomForestClassifier
```

```python
rf = RandomForestClassifier(featuresCol="features", labelCol="Outcome")
model_rf = rf.fit(train_df)
rf_train_prediction = model_rf.transform(train_df)
rf_test_prediction = model_rf.transform(test_df)
#
```

Model Evaluation


```python
from pyspark.mllib.evaluation import MulticlassMetrics
```

```python
lr_train_prediction_casted = lr_train_prediction.withColumn("Outcome", col("Outcome").cast("Double"))
lr_train_prediction_casted.show()
```

```python
from pyspark.ml.evaluation import MulticlassClassificationEvaluator
```

```python
Accuracy = MulticlassClassificationEvaluator(labelCol="Outcome", predictionCol = "prediction",metricName="accuracy")
Accuracy.evaluate(lr_train_prediction)
```

```python
f1 = MulticlassClassificationEvaluator(labelCol="Outcome", predictionCol = "prediction",metricName="f1")
f1.evaluate(lr_train_prediction)
```

```python
precision = MulticlassClassificationEvaluator(labelCol="Outcome", predictionCol = "prediction",metricName="weightedPrecision")
precision.evaluate(lr_train_prediction)
```
